# FlyLM-2 — 저장된 모델로 추론만 하기

`flylm-2-training.ipynb`의 마지막 셀에서 만든 `flylm2_inference.zip`을 풀어 준비한다.
Kaggle에서는 풀린 폴더를 Dataset으로 업로드하고 이 노트북에 연결한다.
`model.pt`, `tokenizer.model`, `tokenizer.json`, `graph/`, `flylm_inference.py`는 같은 폴더 안에 있어야 한다.
학습 데이터와 학습 셀은 필요 없다. 아래에서 폴더 경로를 지정한다.
첫 셀은 P100 미지원 PyTorch만 교체하며, 기존 torch가 로드됐으면 한 번 재시작한다.
CPU 실행은 `DEVICE='cpu'`로 설정한다.

## 뉴런 활동 시각화
전체 165,122개 뉴런을 실제 해부학적 대표 위치에 표시한다.
세포체 위치 140,024개, 공개 골격에서 선택한 실제 점 25,098개를 사용한다.
골격 대표점은 가지가 모인 중심에 가까운 실제 골격 정점이며, 세포체의 위치를 뜻하지 않는다.
두 시점(X–Z / 90도 회전 X–Y)의 밝기와 재생 시간은 동기화된다.
좌표가 노트북에 포함되어 있어 시각화용 추가 다운로드와 재학습은 필요 없다.
입력은 청록색, 내부는 파란색, 출력은 주황색이다. 밝기는 모델의 시냅스 상태 `s`이며 실제 스파이크 기록이 아니다.
HTML은 추론 완료 후 재생하며, 별도 파일로 저장해 브라우저에서 다시 열 수 있다.
자료: [MaleCNS v1.0 / CC-BY](https://male-cns.janelia.org/download/).

재생은 12fps 고정입니다. MP4 저장 버튼은 지도 밖에 있으며 뉴런 지도만 영상으로 저장합니다.

## FlyLM-2 호환 조건

SentencePiece 어휘·특수 토큰·토크나이저 해시와 신경 활동 배정은 저장된 번들에서 복원한다.
토크나이저를 다시 학습하지 않는다. 이전 바이트 모델 체크포인트는 호환되지 않는다.
신경 상태는 `[4,B,N]`: 막전위, 시냅스 신호, 적응, 남은 불응기이다.
미확인 유형을 포함한 신경 동역학은 체크포인트와 함께 저장된 배정을 그대로 복원한다.
미확인 유형의 spiking 기본값을 지원하며, 문헌상 graded 유형은 저장된 연속 배정을 유지한다.
화면의 밝기는 두 방식 모두의 시냅스 상태 `s`이며, 스파이크 횟수나 생체 기록이 아니다.
출력 길이는 바이트 수가 아닌 `MAX_NEW_TOKENS`로 지정한다.

기존 번들에 graded 전용 검증 함수가 남아 있어도 로드 셀에서 호환 검증으로 교체한다.
기존 graded 학습본의 신경 배정을 추론 시 임의로 바꾸지는 않는다.

````python
from pathlib import Path
import importlib.util
import torch

BUNDLE_DIR = Path('/kaggle/input/datasets/chika1472/flylm-2')  # 압축을 푼 실제 내부 폴더로 수정한다.
DEVICE = None  # GPU가 있으면 cuda, 없으면 cpu. 강제 CPU는 'cpu'.

required_files = ['model.pt', 'tokenizer.model', 'flylm_inference.py',
                  'graph/dynamics_ids.npy', 'graph/dynamics_evidence.json']
missing = [name for name in required_files if not (BUNDLE_DIR / name).is_file()]
if missing:
    raise FileNotFoundError(f'BUNDLE_DIR를 번들 내부 폴더로 바꾸세요: {BUNDLE_DIR}; 누락: {missing}')
saved = torch.load(BUNDLE_DIR / 'model.pt', map_location='cpu', weights_only=True)
if saved.get('architecture') != 'malecns_mixed_sentencepiece_v2':
    raise ValueError('SentencePiece를 사용하는 FlyLM-2 체크포인트가 필요합니다.')
del saved
module_path = BUNDLE_DIR / 'flylm_inference.py'
spec = importlib.util.spec_from_file_location('saved_flylm2', module_path)
saved_flylm = importlib.util.module_from_spec(spec)
spec.loader.exec_module(saved_flylm)

# 기존 Kaggle 번들의 graded 전용 검증 함수를 호환 가능한 검증으로 교체한다.
# 체크포인트 설정·동역학 배정·해시 검사는 유지한다.
import math

def validate_dynamics_config(cfg):
    if cfg['neuron_mode'] not in ('mixed', 'graded_control'):
        raise ValueError('neuron_mode은 mixed 또는 graded_control이어야 한다.')
    if cfg['unknown_activity'] not in ('graded', 'spiking'):
        raise ValueError('미확인 유형의 기본 동역학은 graded 또는 spiking이어야 한다.')
    for key in ('spike_threshold_center', 'spike_threshold_radius', 'spike_reset', 'surrogate_slope'):
        if not math.isfinite(cfg[key]):
            raise ValueError(f'유한한 동역학 설정이 필요하다: {key}')
    if (cfg['spike_threshold_radius'] < 0 or cfg['surrogate_slope'] <= 0
            or cfg['spike_threshold_center'] - cfg['spike_threshold_radius'] <= max(0, cfg['spike_reset'])):
        raise ValueError('스파이크 역치는 항상 0과 리셋 전압보다 높아야 한다.')
    if (type(cfg['refractory_ticks']) is not int or cfg['refractory_ticks'] < 0
            or type(cfg['brain_steps']) is not int or cfg['brain_steps'] < 1):
        raise ValueError('불응기는 0 이상의 정수, 토큰당 tick 수는 양의 정수여야 한다.')

saved_flylm.validate_dynamics_config = validate_dynamics_config
model = saved_flylm.load_model(BUNDLE_DIR / 'model.pt', device=DEVICE)
print('모델 로드 완료:', model.embedding.weight.device)
print('SentencePiece 어휘:', model.vocab_size, '/ 스파이크 동역학 뉴런:', int(model.spiking_mask.sum()))
````

## 1. 시각화 함수

이 셀은 저장된 모델의 `advance()` 결과를 관찰한다. 가중치·학습·토큰 선택은 바꾸지 않는다.
한 frame은 한 SentencePiece 토큰 처리가 끝난 뒤의 상태이며, 기록 간격은 `MAX_FRAMES`에 맞춰 자동 조절한다.
추론 자체는 모든 SentencePiece 토큰를 계산하고, 화면에는 **모든 뉴런**을 유지한다.

기록 함수는 네 상태를 그대로 전달하고 `s`만 관찰한다. 텍스트 표시는 완성된 UTF-8 문자까지만 복원한다.


````python
"""실제 FlyLM-2 상태를 기록하고 모든 뉴런을 오프라인 2D 화면에 표시한다."""
import base64
import codecs
import gzip
import html
import json
import math
from pathlib import Path
import urllib.request

import numpy as np
import torch

ANNOTATION_URL = (
    'https://storage.googleapis.com/flyem-male-cns/v1.0/connectome-data/flat-connectome/'
    'body-annotations-male-cns-v1.0-minconf-0.5.feather'
)


def neuron_layout(model, graph_dir, cache_dir, annotation_path=None, coordinate_path=None):
    """공식 세포체·골격 대표점을 뉴런 ID로 연결하고 좌표 출처를 유지한다."""
    graph_dir, cache_dir = Path(graph_dir), Path(cache_dir)
    cache_dir.mkdir(parents=True, exist_ok=True)
    body_ids = np.load(graph_dir / 'body_ids.npy', allow_pickle=False)
    if body_ids.shape != (model.n,) or len(np.unique(body_ids)) != model.n:
        raise ValueError('body_ids.npy does not match the model neuron count')
    if coordinate_path is not None:
        with np.load(coordinate_path, allow_pickle=False) as saved:
            metadata = json.loads(str(saved['metadata']))
            if metadata['dataset'] != 'male-cns:v1.0' or metadata['units'] != '8nm_voxels':
                raise ValueError('Coordinate dataset/units do not match MaleCNS v1.0')
            ids, coordinates, sources = saved['body_ids'], saved['xyz'], saved['source']
            if (coordinates.shape != (len(ids), 3) or sources.shape != (len(ids),)
                    or len(np.unique(ids)) != len(ids) or not np.isin(sources, [0, 1, 2, 3]).all()):
                raise ValueError('Invalid coordinate table')
            order = np.argsort(ids)
            positions = np.searchsorted(ids[order], body_ids)
            if (positions >= len(ids)).any() or not np.array_equal(ids[order][positions], body_ids):
                raise ValueError('Coordinate table is missing model body IDs')
            xyz = coordinates[order[positions]].copy()
            coordinate_kind = sources[order[positions]].astype(np.uint8)
            if not np.array_equal(np.isfinite(xyz).all(1), coordinate_kind > 0):
                raise ValueError('Coordinate validity and provenance disagree')
    else:
        import pyarrow.feather as feather

        if annotation_path is not None and not Path(annotation_path).is_file():
            raise FileNotFoundError(f'Annotation file not found: {annotation_path}')
        annotation_path = Path(annotation_path) if annotation_path else cache_dir / 'body-annotations.feather'
        if not annotation_path.exists():
            print('Downloading official soma annotations (about 13 MB)...', flush=True)
            temporary = annotation_path.with_suffix('.partial')
            urllib.request.urlretrieve(ANNOTATION_URL, temporary)
            temporary.replace(annotation_path)
        table = feather.read_table(annotation_path, columns=['bodyId', 'somaLocation'])
        ids = table['bodyId'].to_numpy()
        order = np.argsort(ids)
        sorted_ids = ids[order]
        locations = table['somaLocation'].to_pylist()
        positions = np.searchsorted(sorted_ids, body_ids)
        found = positions < len(sorted_ids)
        found[found] &= sorted_ids[positions[found]] == body_ids[found]
        xyz = np.full((model.n, 3), np.nan, dtype=np.float32)
        for index in np.flatnonzero(found):
            value = locations[int(order[positions[index]])]
            if value is not None and len(value) == 3 and np.isfinite(value).all():
                xyz[index] = value
        coordinate_kind = np.isfinite(xyz).all(1).astype(np.uint8)
    known = np.isfinite(xyz).all(1)
    if not known.any():
        raise ValueError('No coordinates matched. Check body IDs and the coordinate source.')
    roles = np.zeros(model.n, dtype=np.uint8)
    roles[model.inputs.detach().cpu().numpy()] = 1
    roles[model.outputs.detach().cpu().numpy()] = 2
    xy = np.zeros((model.n, 2), dtype=np.float32)
    xy_front = np.full_like(xy, -2)
    # X축을 중심으로 시점을 90도 회전한다: X–Z → X–Y.
    aspect = 1200 / 560
    centers, view_width = ((0.25, 0.75), 0.43) if known.all() else ((0.19, 0.53), 0.30)
    for coordinates, vertical_axis, center_x in [(xy, 2, centers[0]), (xy_front, 1, centers[1])]:
        projected = xyz[known][:, [0, vertical_axis]]
        span = np.maximum(np.ptp(projected, axis=0), 1)
        scale = min(view_width * aspect / span[0], 0.78 / span[1])
        centered = projected - (projected.min(0) + projected.max(0)) / 2
        coordinates[known] = centered * [scale / aspect, scale] + [center_x, 0.52]
    # 좌표가 없는 뉴런은 별도 영역에 배치한다.
    for role, low, high in [(1, 0.19, 0.39), (0, 0.49, 0.72), (2, 0.82, 0.93)]:
        members = np.flatnonzero(~known & (roles == role))
        if not len(members):
            continue
        columns = max(1, math.ceil(math.sqrt(len(members) * 0.28 / (high - low))))
        rows = math.ceil(len(members) / columns)
        indices = np.arange(len(members))
        xy[members, 0] = 0.76 + 0.21 * ((indices % columns) + 0.5) / columns
        xy[members, 1] = low + (high - low) * ((indices // columns) + 0.5) / rows
    return dict(xy=xy, xy_front=xy_front, roles=roles, body_ids=body_ids, known=known,
                xyz=xyz, coordinate_kind=coordinate_kind,
                coordinate_source=str(coordinate_path) if coordinate_path is not None else ANNOTATION_URL)


@torch.no_grad()
def record_activity(model, runtime, question, max_new_tokens=128, temperature=0.0, max_frames=160):
    """토큰 선택과 혼합 동역학을 바꾸지 않고 실제 추론의 시냅스 상태를 기록한다."""
    if type(max_frames) is not int or max_frames < 4 or type(max_new_tokens) is not int or max_new_tokens < 1:
        raise ValueError('max_frames는 4 이상, max_new_tokens는 1 이상의 정수여야 한다.')
    prefix = runtime.inference_inputs(model, question)
    stride = max(1, math.ceil((len(prefix) + max_new_tokens) / (max_frames - 3)))
    frames = [np.zeros(model.n, dtype=np.uint8)]
    captions = [dict(step=0, phase='ready', input='', output='')]
    input_ids, output_ids = [], []
    step, latest = 0, None
    original = model.advance
    had_override = 'advance' in model.__dict__
    prior_override = model.__dict__.get('advance')

    def decoded(ids):
        # 끝부분 fallback 바이트가 불완전하면 해당 문자만 표시를 미룬다.
        tail = len(ids)
        while tail and model.tokenizer.is_byte(ids[tail - 1]):
            tail -= 1
        pending = codecs.getincrementaldecoder('utf-8')('strict')
        for token in ids[tail:]:
            pending.decode(bytes([int(model.tokenizer.id_to_piece(token)[3:5], 16)]), final=False)
        count = len(pending.getstate()[0])
        return model.tokenizer.decode(ids[:-count] if count else ids)

    def save_frame(state, phase):
        # 시냅스 상태에 고정된 제곱근 밝기 척도를 적용한다.
        signal = state[1, 0]
        if not bool(torch.isfinite(signal).all()):
            raise RuntimeError('뉴런 활동에 유한하지 않은 값이 있다.')
        frames.append(signal.clamp(0, 1).sqrt().mul(255).round().to(torch.uint8).cpu().numpy())
        captions.append(dict(step=step, phase=phase, input=decoded(input_ids), output=decoded(output_ids)))

    def observe(token, *args, **kwargs):
        nonlocal step, latest
        latest = original(token, *args, **kwargs)
        step += 1
        phase = 'input' if step <= len(prefix) else 'output'
        value = int(token[0])
        if value not in (model.bos_id, model.eos_id, model.pad_id):
            (input_ids if phase == 'input' else output_ids).append(value)
        if step % stride == 0 or step == len(prefix):
            save_frame(latest, phase)
        if step % 64 == 0:
            print(f'기록 중: {phase}, {step}토큰 처리', flush=True)
        return latest

    model.advance = observe
    try:
        answer = runtime.reply(model, question, max_new_tokens=max_new_tokens, temperature=temperature)
        save_frame(latest, 'done')
        captions[-1]['output'] = answer
    finally:
        if had_override:
            model.advance = prior_override
        else:
            del model.advance
    return dict(activity=np.stack(frames), captions=captions, question=question,
                answer=answer, stride=stride, brain_steps=model.k)


PLAYER_HTML = r'''<!doctype html>
<html lang="ko"><head><meta charset="utf-8"><meta name="viewport" content="width=device-width,initial-scale=1">
<title>FlyLM-2 · 뉴런 활동</title>
<style>
:root{--canvas:#fff;--ink:#171717;--body:#60646c;--line:#dcdee0;--dark:#171717;--primary:#000;--active:#1a1a1a}
*{box-sizing:border-box}body{margin:0;background:var(--canvas);color:var(--ink);font:14px/1.55 Inter,-apple-system,system-ui,sans-serif}
main{max-width:1280px;margin:auto;padding:24px}
.legend{display:flex;gap:20px;flex-wrap:wrap;font-size:12px;color:var(--body)}.dot{display:inline-block;width:8px;height:8px;border-radius:50%;margin-right:6px}
.stage{position:relative;margin:16px 0;border:1px solid var(--line);border-radius:12px;overflow:hidden;background:var(--dark)}
canvas{display:block;width:100%;height:560px}
.tray-label{position:absolute;left:75%;color:#b0b4ba;font-size:11px;pointer-events:none}.divider{position:absolute;left:72%;top:9%;bottom:5%;border-left:1px dashed #444;pointer-events:none}
#error{position:absolute;inset:25%;background:var(--dark);color:#eb8e90;padding:24px;display:none}
button{height:40px;padding:10px 18px;border:1px solid var(--primary);border-radius:8px;background:var(--primary);color:#fff;font:500 14px Inter,system-ui,sans-serif;cursor:pointer;min-width:85px}
button:hover{background:var(--active)}button:disabled{opacity:.4;cursor:default}button:focus-visible{outline:2px solid var(--ink);outline-offset:3px}
.export-tools{display:flex;align-items:center;gap:12px;margin-top:24px;padding-top:16px;border-top:1px solid #f0f0f3}
#export{background:var(--canvas);color:var(--ink);border-color:var(--line)}#export:hover{background:#fafafa}#exportStatus{color:var(--body);font-size:13px}
.texts{display:grid;grid-template-columns:1fr 1fr;gap:16px;margin-top:24px}.text{min-height:100px;background:#fafafa;border:1px solid #f0f0f3;border-radius:12px;padding:16px}
.text label{font-size:12px;color:var(--body)}.text pre{white-space:pre-wrap;word-break:break-word;font:14px/1.65 Inter,system-ui,sans-serif;margin:8px 0 0}
#notice{color:#ab6400;font-size:13px}
@media(max-width:650px){main{padding:16px}canvas{height:440px}.texts{grid-template-columns:1fr}.legend{gap:12px}.tray-label{font-size:9px}.export-tools{flex-wrap:wrap}}
</style></head><body><main>
<div id="notice"></div>
<div class="legend"><span><i class="dot" style="background:#5ce6f3"></i>입력 · 감각 뉴런</span><span><i class="dot" style="background:#8399fc"></i>내부 뉴런</span><span><i class="dot" style="background:#ffb674"></i>출력 · 운동 뉴런</span></div>
<div class="stage"><canvas id="brain" aria-label="전체 뉴런의 2D 활동 지도"></canvas>
<div class="divider" id="viewDivider" style="left:36%"></div><div class="divider" id="trayDivider"></div>
<div class="tray-label" style="top:3%">좌표 없는 뉴런 · 별도 배열</div>
<div class="tray-label" style="top:14%" id="missingInput"></div><div class="tray-label" style="top:44%" id="missingInternal"></div><div class="tray-label" style="top:77%" id="missingOutput"></div><div id="error" role="alert"></div></div>
<div class="texts"><div class="text"><label>읽고 있는 입력</label><pre id="inputText"></pre></div><div class="text"><label>생성 중인 답변</label><pre id="outputText"></pre></div></div>
<div class="export-tools" aria-label="영상 저장"><button id="export" disabled>MP4 저장</button><span id="exportStatus" role="status" aria-live="polite"></span></div>
</main><script id="payload" type="application/json">__PAYLOAD__</script><script>
(async()=>{
const d=JSON.parse(document.getElementById('payload').textContent), $=id=>document.getElementById(id);
const fail=message=>{ $('error').style.display='block';$('error').textContent=message;throw new Error(message); };
async function unpack(encoded){
 const raw=atob(encoded),bytes=new Uint8Array(raw.length);for(let i=0;i<raw.length;i++)bytes[i]=raw.charCodeAt(i);
 if(typeof DecompressionStream==='undefined')fail('이 브라우저는 압축된 활동 데이터를 지원하지 않습니다. 최신 Chrome 또는 Edge에서 HTML을 여세요.');
 return new Response(new Blob([bytes]).stream().pipeThrough(new DecompressionStream('gzip'))).arrayBuffer();
}
const [xyRaw,frontRaw,roleRaw,activityRaw]=await Promise.all([unpack(d.xy),unpack(d.xy_front),unpack(d.roles),unpack(d.activity)]);
const xy=new Float32Array(xyRaw),front=new Float32Array(frontRaw),roles=new Uint8Array(roleRaw),activity=new Uint8Array(activityRaw);
const n=d.neurons,total=d.captions.length,canvas=$('brain');
if(xy.length!==n*2||front.length!==n*2||roles.length!==n||activity.length!==n*total)fail('활동 데이터 크기가 일치하지 않습니다.');
const gl=canvas.getContext('webgl',{alpha:true,antialias:false,preserveDrawingBuffer:true});
if(!gl)fail('WebGL을 사용할 수 없습니다. HTML을 Chrome/Edge에서 열고 하드웨어 가속을 확인하세요.');
function shader(type,source){const s=gl.createShader(type);gl.shaderSource(s,source);gl.compileShader(s);if(!gl.getShaderParameter(s,gl.COMPILE_STATUS))fail(gl.getShaderInfoLog(s));return s;}
const program=gl.createProgram();
gl.attachShader(program,shader(gl.VERTEX_SHADER,`
 attribute vec2 position;attribute float role;attribute float activity;
 uniform float gain;uniform float pixelRatio;varying vec3 color;varying float light;
 void main(){gl_Position=vec4(position.x*2.0-1.0,1.0-position.y*2.0,0.,1.);
 light=clamp(activity*gain,0.0,1.0);color=role>1.5?vec3(1.,.71,.45):(role>.5?vec3(.36,.90,.95):vec3(.51,.60,.99));
 gl_PointSize=pixelRatio*(1.15+3.5*light);}`));
gl.attachShader(program,shader(gl.FRAGMENT_SHADER,`
 precision mediump float;varying vec3 color;varying float light;
 void main(){vec2 p=gl_PointCoord*2.-1.;float r=dot(p,p);if(r>1.)discard;
 float alpha=exp(-3.*r)*(.05+.64*light);gl_FragColor=vec4(mix(color,vec3(1.),light*.3),alpha);}`));
gl.linkProgram(program);if(!gl.getProgramParameter(program,gl.LINK_STATUS))fail(gl.getProgramInfoLog(program));gl.useProgram(program);
function attribute(name,data,size,type,normalized){const b=gl.createBuffer();gl.bindBuffer(gl.ARRAY_BUFFER,b);gl.bufferData(gl.ARRAY_BUFFER,data,gl.DYNAMIC_DRAW);const loc=gl.getAttribLocation(program,name);gl.enableVertexAttribArray(loc);gl.vertexAttribPointer(loc,size,type,normalized,0,0);return b;}
const positionBuffer=attribute('position',xy,2,gl.FLOAT,false);
attribute('role',roles,1,gl.UNSIGNED_BYTE,false);
const activityBuffer=attribute('activity',activity.subarray(0,n),1,gl.UNSIGNED_BYTE,true);
const gainLocation=gl.getUniformLocation(program,'gain'),ratioLocation=gl.getUniformLocation(program,'pixelRatio');
gl.enable(gl.BLEND);gl.blendFunc(gl.SRC_ALPHA,gl.ONE);gl.clearColor(0,0,0,0);
const FPS=12, GAIN=3;
let frame=0,playing=false,last=0,exporting=false;
function draw(){
 const ratio=Math.min(window.devicePixelRatio||1,2),w=Math.round(canvas.clientWidth*ratio),h=Math.round(canvas.clientHeight*ratio);
 if(canvas.width!==w||canvas.height!==h){canvas.width=w;canvas.height=h;}gl.viewport(0,0,w,h);
 gl.uniform1f(gainLocation,GAIN);gl.uniform1f(ratioLocation,ratio);
 gl.bindBuffer(gl.ARRAY_BUFFER,activityBuffer);gl.bufferSubData(gl.ARRAY_BUFFER,0,activity.subarray(frame*n,(frame+1)*n));
 gl.clear(gl.COLOR_BUFFER_BIT);
 gl.bindBuffer(gl.ARRAY_BUFFER,positionBuffer);
 for(const positions of [xy,front]){gl.bufferSubData(gl.ARRAY_BUFFER,0,positions);gl.drawArrays(gl.POINTS,0,n);}
 const c=d.captions[frame];
 $('inputText').textContent=c.input||'입력을 기다리는 중';$('outputText').textContent=c.output||(c.phase==='done'?'(빈 답변)':'…');
}
function pause(){playing=false;}
new ResizeObserver(draw).observe(canvas);
function tick(now){if(playing&&now-last>=1000/FPS){frame=Math.min(total-1,frame+1);last=now;draw();if(frame===total-1)pause();}requestAnimationFrame(tick);}
if(d.known===n){$('viewDivider').style.left='50%';$('trayDivider').style.display='none';document.querySelectorAll('.tray-label').forEach(e=>e.style.display='none');}
$('missingInput').textContent=`감각 · ${d.missing[1].toLocaleString()}개`;$('missingInternal').textContent=`내부 · ${d.missing[0].toLocaleString()}개`;$('missingOutput').textContent=`운동 · ${d.missing[2].toLocaleString()}개`;

// 뉴런 지도와 동기화된 입력·출력 카드를 저장하고 조작 버튼은 제외한다.
$('export').onclick=async()=>{
 if(exporting)return;
 const status=$('exportStatus');
 const mime=typeof MediaRecorder!=='undefined' && ['video/mp4;codecs=avc1.42001E','video/mp4'].find(t=>MediaRecorder.isTypeSupported(t));
 if(!mime){status.textContent='이 브라우저는 MP4 저장을 지원하지 않습니다. 최신 Chrome/Edge에서 HTML 파일을 열어 주세요.';return;}
 exporting=true;const savedPlaying=playing;pause();const savedFrame=frame;
 $('export').disabled=true;
 let stream,recorder;
 try{
  const video=document.createElement('canvas');
  video.width=Math.max(2,Math.floor(canvas.width/2)*2);
  const brainHeight=Math.max(2,Math.floor(canvas.height/2)*2);
  const ctx=video.getContext('2d',{alpha:false});
  const scale=video.width/Math.max(1,canvas.clientWidth),pad=16*scale,gap=16*scale;
  const stacked=canvas.clientWidth<=650;
  const cardWidth=stacked?video.width-2*pad:(video.width-2*pad-gap)/2;
  const font=`${14*scale}px Inter, system-ui, sans-serif`,lineHeight=24*scale;
  // 자막 높이를 먼저 계산해 영상 크기를 고정하고 잘림을 방지한다.
  const wrap=text=>{
   ctx.font=font;const lines=[];
   for(const paragraph of text.split('\n')){
    let line='';
    for(const char of Array.from(paragraph)){
     if(line&&ctx.measureText(line+char).width>cardWidth-2*pad){lines.push(line);line='';}
     line+=char;
    }
    lines.push(line);
   }
   return lines;
  };
  const captionText=c=>[c.input||'입력을 기다리는 중',c.output||(c.phase==='done'?'(빈 답변)':'…')];
  const captionLines=d.captions.map(c=>captionText(c).map(wrap));
  const maxLines=captionLines.reduce((max,pair)=>Math.max(max,...pair.map(lines=>lines.length)),1);
  const cardHeight=pad*2+28*scale+maxLines*lineHeight;
  video.height=Math.ceil((brainHeight+pad*2+cardHeight*(stacked?2:1)+(stacked?gap:0))/2)*2;
  const paint=()=>{
   ctx.fillStyle='#ffffff';ctx.fillRect(0,0,video.width,video.height);
   ctx.fillStyle='#171717';ctx.fillRect(0,0,video.width,brainHeight);
   ctx.drawImage(canvas,0,0,video.width,brainHeight);
   ['읽고 있는 입력','생성 중인 답변'].forEach((title,index)=>{
    const x=pad+(stacked?0:index*(cardWidth+gap));
    const y=brainHeight+pad+(stacked?index*(cardHeight+gap):0);
    ctx.beginPath();ctx.roundRect(x,y,cardWidth,cardHeight,12*scale);
    ctx.fillStyle='#fafafa';ctx.fill();ctx.strokeStyle='#f0f0f3';ctx.lineWidth=scale;ctx.stroke();
    ctx.textBaseline='top';ctx.fillStyle='#60646c';ctx.font=`${12*scale}px Inter, system-ui, sans-serif`;
    ctx.fillText(title,x+pad,y+pad);
    ctx.fillStyle='#171717';ctx.font=font;
    captionLines[frame][index].forEach((line,i)=>ctx.fillText(line,x+pad,y+pad+28*scale+i*lineHeight));
   });
  };
  frame=0;draw();paint();stream=video.captureStream(FPS);
  recorder=new MediaRecorder(stream,{mimeType:mime,videoBitsPerSecond:8000000});
  const chunks=[];let recordingError=null;
  recorder.ondataavailable=e=>{if(e.data.size)chunks.push(e.data);};
  const stopped=new Promise(resolve=>{recorder.onstop=resolve;});
  recorder.onerror=e=>{recordingError=e.error||new Error('MP4 인코딩 실패');};
  recorder.start();
  for(let i=0;i<total;i++){
   if(recordingError)throw recordingError;
   frame=i;draw();paint();status.textContent=`MP4 저장 중 · ${i+1} / ${total}`;
   await new Promise(resolve=>setTimeout(resolve,1000/FPS));
  }
  if(recorder.state!=='inactive')recorder.stop();await stopped;
  if(recordingError)throw recordingError;
  const blob=new Blob(chunks,{type:'video/mp4'});
  if(!blob.size)throw new Error('영상 데이터가 비어 있습니다.');
  const url=URL.createObjectURL(blob),a=document.createElement('a');
  a.href=url;a.download='flylm2-activity.mp4';document.body.appendChild(a);a.click();a.remove();
  setTimeout(()=>URL.revokeObjectURL(url),60000);status.textContent='MP4 저장 완료';
 }catch(e){status.textContent=`MP4 저장 실패: ${e.message}`;}
 finally{
  if(recorder&&recorder.state!=='inactive')recorder.stop();
  if(stream)stream.getTracks().forEach(t=>t.stop());
  frame=savedFrame;draw();playing=savedPlaying;last=performance.now();exporting=false;$('export').disabled=false;
 }
};

$('notice').textContent=d.notice||'';$('export').disabled=false;draw();playing=true;last=performance.now();requestAnimationFrame(tick);
})().catch(e=>{const el=document.getElementById('error');el.style.display='block';el.textContent=e.message;console.error(e);});
</script></body></html>'''


def activity_html(trace, layout, notice=''):
    def pack(array):
        return base64.b64encode(
            gzip.compress(np.ascontiguousarray(array).tobytes(), compresslevel=3)
        ).decode('ascii')

    n = len(layout['roles'])
    if trace['activity'].shape != (len(trace['captions']), n):
        raise ValueError('Activity shape does not match neuron layout')
    payload = dict(
        neurons=n,
        known=int(layout['known'].sum()),
        location_counts=[
            int((layout['coordinate_kind'] == code).sum())
            for code in range(4)
        ],
        missing=[
            int((~layout['known'] & (layout['roles'] == role)).sum())
            for role in range(3)
        ],
        xy=pack(layout['xy'].astype('<f4')),
        xy_front=pack(layout['xy_front'].astype('<f4')),
        roles=pack(layout['roles'].astype('u1')),
        activity=pack(trace['activity'].astype('u1')),
        captions=trace['captions'],
        stride=trace['stride'],
        brain_steps=trace['brain_steps'],
        notice=notice,
    )
    # 사용자 텍스트는 JSON과 textContent로만 전달한다.
    data = json.dumps(payload, ensure_ascii=False).replace('<', '\\u003c')
    return PLAYER_HTML.replace('__PAYLOAD__', data)


def show_activity(trace, layout, output_path):
    from IPython.display import HTML, display

    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)
    document = activity_html(trace, layout)
    output_path.write_text(document, encoding='utf-8')
    print('오프라인 활동 화면:', output_path)
    print('답변:', trace['answer'])
    display(HTML(
        '<iframe sandbox="allow-scripts allow-downloads" '
        'style="width:100%;height:900px;border:0" '
        'title="FlyLM neuron activity" srcdoc="'
        + html.escape(document, quote=True)
        + '"></iframe>'
    ))
    return output_path
````

````python
# 오프라인 시각화를 위한 공식 MaleCNS v1.0 좌표를 포함한다.
# 출처 코드 1: 세포체, 2: 실제 SWC 골격 정점, 3: tosoma 주석.
import base64
import hashlib
from pathlib import Path

COORDINATE_SHA256 = '8d3602f8ffdfc30239d98a4c484e59a313bff9ea8558ce718af18037b90c58ca'
COORDINATE_DATA = ''

def bundled_coordinates(cache_dir):
    cache_dir = Path(cache_dir)
    cache_dir.mkdir(parents=True, exist_ok=True)
    path = cache_dir / f'neuron_coordinates_{COORDINATE_SHA256[:12]}.npz'
    if not path.exists() or hashlib.sha256(path.read_bytes()).hexdigest() != COORDINATE_SHA256:
        raw = base64.b64decode(COORDINATE_DATA)
        if hashlib.sha256(raw).hexdigest() != COORDINATE_SHA256:
            raise ValueError('Bundled coordinate data failed SHA-256 verification')
        temporary = path.with_suffix('.partial')
        temporary.write_bytes(raw)
        temporary.replace(path)
    return path
````
